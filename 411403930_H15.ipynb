題目：請設計讓使用者輸入A倉與B倉的盤點文具清單，輸入 '/' 表示結束，並將二倉列出所有清單名稱並用斜線 / 分隔符號列出，最後加上“膠水”項目。

輸出範例：
請輸入A倉文具，輸入'/'結束：
鉛筆
橡皮擦
稿紙
/
請輸入B倉文具，輸入'/'結束：
文件夾
訂書機
訂書針
/
二倉合併清單：
鉛筆/橡皮擦/稿紙/文件夾/訂書機/訂書針/膠水



```python
listA=[]
while True:
  item=input("請輸入A倉文具，輸入'/'結束")
  if item=="/":
    break
  listA.append(item)
listB=[]
while True:
  item=input("請輸入B倉文具，輸入'/'結束")
  if item=="/":
    break
  listB.append(item)
listC=listA+listB
listC.append("膠水")
listC="/".join(listC)
print(f"二倉合併清單：{listC}")
```
