```python
def add(p,q,dis=0.05):
  total=p*q
  tax=total*dis
  total_all=total+tax
  return total,tax,total_all

p=int(input("請輸入單價: "))
q=int(input("請輸入數量: "))

x,y,z=add(p,q)
print(f"小計:{x}\n稅金:{y}\n總金額:{z}")
```

題目：請利用定義函式傳回多筆方式，設計讓使用者輸入購物金額以及購物數量，並計算稅金（稅金=小計×5%），再輸出小計、稅金、總金額。



輸入範例：

請輸入單價：100

請輸入數量：3



輸出範例：

稅金: 15.0

總金額: 315.0
