```python
import random

day=1
max_day=3
win=False
while day<=max_day and not win:
  lucky=random.randint(1,50)
  guess=int(input(f"今天第{day}天請輸入1-50數字"))
  print(f"今天的幸運數字是{lucky}")
  if guess == lucky:
    print("你中獎了")
    win=True
  else:
    print("你沒中獎，明天再來")
    day+=1
if not win:
    print("很遺憾3天都沒中獎，下次再試試吧")
```

題目：隨機抽獎3天模擬

公司要辦抽獎活動3天，每天會抽出一個幸運號碼（1~50）。

使用者每天可以輸入自己想抽的號碼（1~50）。

如果號碼和當天的幸運號碼相同，就中獎，程式結束。

如果不相同，就顯示「沒中，明天再來！」並繼續抽下一天。

程式要顯示每一天抽到的幸運號碼，並至3天即截止。

輸出範例：

今天第1天,請輸入1-50數字5

今天的幸運數字是2

你沒中獎，明天再來



今天第2天,請輸入1-50數字1

今天的幸運數字是14

你沒中獎，明天再來



今天第3天,請輸入1-50數字26

今天的幸運數字是30

你沒中獎，明天再來

很遺憾3天都沒中獎，下次再試試吧！
