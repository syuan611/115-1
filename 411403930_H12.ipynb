```python
def f1(p,q):
  return p*q
def f2(p,q):
    total=f1(p,q)
    return total*0.9
p=int(input("請輸單價: "))
q=int(input("請輸數量: "))
print(f1(p,q))
print(f2(p,q))
```

題目：利用副程式呼叫副程式讓使用者輸入p為單價，q為數量，計算原價總金額與打9折後的金額。



輸出範例：

原價金額：200元

9折金額：180元
