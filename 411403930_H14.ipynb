```python
def h(n):
  if n==1:
    return 1
  if n==2:
    return 2
  return h(n-1)+h(n-2)
sum=h(10)
print(f"總共有{sum}種方式")
```

題目： 有一隻兔子要跳上總共 10 階的樓梯，每次只能跳 1 階或 2 階。請使用 Python 設計一個名為 f(n) 的遞迴函式，計算兔子跳到第 10 階總共有幾種不同的跳法。已知跳上第 1 階有 1 種方法，跳上第 2 階有 2 種方法。請利用遞迴完成計算，不可使用 for 或 while 迴圈。


輸出範例：
總共有89種方式
