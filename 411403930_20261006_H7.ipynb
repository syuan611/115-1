```python
import math
r=float(input("請輸入角度（度）:"))
a=r*(math.pi/180)
b=math.sin(a)
print(f"你計算的sin值為:{b:.2f}")
```

請使用者輸入一個角度（單位為度），將其轉換為弧度後，計算並輸出該角度的 sin 值（取小數點後兩位）。公式：弧度 = 角度 × (π / 180)



輸入範例：

請輸入角度（度）：30

輸出範例：

你計算的sin值為:0.50
