```python
import random
while True:
  s=random.randint(1,5)
  if s==1:
    print("遇到怪物")
  elif s==2:
    print("找到寶物")
  elif s==3:
    print("陷阱跳過")
    continue
  elif s==4:
    print("補充生命")
  elif s==5:
    print("出口結束")
    break
```

題目：請設計一個隨機遊戲事件，事件規則如下：

1 → 遇到怪物

2 → 找到寶物

3 → 陷阱跳過（continue）

4 → 補充生命

5 → 出口結束（break）



輸出範例：

找到寶物
