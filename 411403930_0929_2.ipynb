```python
x=int(input("請輸入年齡: "))
y=int(input("請輸入年齡: "))
a=x>12 and x<65
b=y>12 and y<65
print("是否需要門票:",a)
print("是否需要門票:",b)
```

題目：寫一個程式僅用關係運算子方式，請使用者輸入遊客年齡，判斷是否要門票，判斷規則如：12歲以下兒童，65歲以上老人免門票，其餘均需門票。


輸入範例：
請輸入年齡：66
請輸入年齡：20


輸出範例：
是否需門票：False
是否需門票：True
