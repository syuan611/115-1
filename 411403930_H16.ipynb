題目：請設計讓使用者輸入點餐清單，點餐完輸入 '/' 表示結束，但若訂單中有 "沙拉"，自動刪除並提示「今日無沙拉，已自動移除。」最後訂單贈送冰淇淋，並列出訂單內容。



輸入範例：

請輸入您要點的餐點，輸入'/'結束：

漢堡

可樂

薯條

沙拉

/

輸出範例：

今日無沙拉，已自動移除。

完整訂單清單：

漢堡, 可樂, 薯條, 冰淇淋

```python
order=[]
print("請輸入您要點的餐點，輸入'/'結束：")
while True:
  item=input()
  if item=="/":
    break
  order.append(item)
  if item=="沙拉":
    order.remove("沙拉")
    print("今日無沙拉，已自動移除。")
order.append("冰淇淋")
order=",".join(order)
print(f"完整訂單清單：{order}")
```
